```python
import random


def Matrix_multip(M1, M2):
    result = [[0 for j in range(len(M2[0]))] for i in range(len(M1))]
    for i in range(len(M1)):
        for j in range(len(M2[0])):
            for k in range(len(M2)):
                result[i][j] = result[i][j] + M1[i][k] * M2[k][j]
    return result


rng = random.Random(20261013)
M1 = [[rng.randint(0, 50) for _ in range(10)] for _ in range(5)]
M2 = [[rng.randint(0, 50) for _ in range(5)] for _ in range(10)]

if __name__ == '__main__':
    for name, matrix in [('M1', M1), ('M2', M2), ('M1 * M2', Matrix_multip(M1, M2))]:
        print(name)
        for row in matrix:
            print(row)
```

```python
def Pascal_triangle(k):
    row = [1]
    for _ in range(1, k):
        row = [1] + [row[j] + row[j + 1] for j in range(len(row) - 1)] + [1]
    print(*row)
    return row


if __name__ == '__main__':
    for k in (100, 200):
        print(f'第 {k} 行：')
        Pascal_triangle(k)
```

```python
import random


def Least_moves(x):
    moves = 0
    while x > 1:
        if x % 2 == 0:
            x //= 2
        else:
            x -= 1
        moves += 1
    print(moves)
    return moves


if __name__ == '__main__':
    rng = random.Random(20261013)
    for x in [1, 2, 5] + [rng.randint(1, 100) for _ in range(5)]:
        print(f'x = {x}: ', end='')
        Least_moves(x)
```

```python
import random
from pathlib import Path
import matplotlib.pyplot as plt


def Expression_table():
    digits = '123456789'
    dp = [{0: ['']}] + [{} for _ in range(9)]
    for end in range(1, 10):
        for start in range(end):
            part = digits[start:end]
            number = int(part)
            for value, expressions in dp[start].items():
                for expression in expressions:
                    if start == 0:
                        dp[end].setdefault(number, []).append(part)
                    else:
                        dp[end].setdefault(value + number, []).append(expression + '+' + part)
                        dp[end].setdefault(value - number, []).append(expression + '-' + part)
    return dp[9]


expressions = Expression_table()
Total_solutions = [len(expressions.get(i, [])) for i in range(1, 101)]


def Find_expression(x):
    solutions = expressions.get(x, [])
    for expression in solutions:
        print(f'{expression} = {x}')
    return solutions


def Plot_solutions():
    plt.rcParams['font.sans-serif'] = ['Arial Unicode MS', 'SimHei', 'DejaVu Sans']
    plt.rcParams['axes.unicode_minus'] = False
    fig, ax = plt.subplots(figsize=(8, 3.2))
    ax.plot(range(1, 101), Total_solutions, color='#2864a5', linewidth=1)
    ax.set(xlabel='目标整数', ylabel='表达式数量', xlim=(1, 100), ylim=(0, max(Total_solutions) + 2))#建议后续中文用英文表达，不然notebook会乱码
    ax.spines[['top', 'right']].set_visible(False)
    fig.tight_layout()
    Path('assets').mkdir(exist_ok=True)
    fig.savefig('assets/total_solutions.png', dpi=180)
    return fig


if __name__ == '__main__':
    x = random.Random(20261013).randint(1, 100)
    print(f'随机目标：{x}')
    Find_expression(x)
    print('Total_solutions =', Total_solutions)
    for label, value in [('最大', max(Total_solutions)), ('最小', min(Total_solutions))]:
        targets = [i for i, count in enumerate(Total_solutions, 1) if count == value]
        print(f'{label}值：{value}，对应整数：{targets}')
    Plot_solutions()
```
